```python
import matplotlib.pyplot as plt
import numpy as np

pressures_as = [0, 2, 4, 6]
shannon_as = [5.870391, 5.867458, 5.862750, 5.855970]

pressures_p = [0, 1.38, 3.61, 6.3]
shannon_p = [5.323853, 5.337695, 5.360069, 5.591013]
shannon_p_std = [0.013575, 0.006972, 0.011370, 0.029578]

fig, ax = plt.subplots(figsize=(2.4, 2.4))

ax.plot(
    pressures_p,
    shannon_p,
    marker='o',
    linestyle='--',
    dashes=[2,1],
    color='#7c87b5',
    markersize=5,
    label=r'$a$-P',
)

ax.plot(
    pressures_as,
    shannon_as,
    marker='o',
    linestyle='-',
    color='#cc5d5d',
    markersize=5,
    label=r'$a$-As',
)

ax.set_xlabel("Pressure (GPa)")
ax.set_ylabel("Entropy (bits)")
ax.set_xlim(-1, 7)
ax.set_xticks(np.arange(0, 6.01, 2))
ax.set_ylim(5.1, 6.1)
ax.set_yticks(np.arange(5.1, 6.11, 0.2))
ax.legend()
ax.yaxis.set_tick_params(which='both', labelbottom=True, bottom=True, zorder=10)
ax.xaxis.set_tick_params(which='both', labelbottom=True, bottom=True, zorder=10)

plt.show()
```
