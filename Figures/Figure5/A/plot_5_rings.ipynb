```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("rings_5.csv")

df_as = df[df["system"] == "a-As"]
df_p  = df[df["system"] == "a-P"]

labels_as = df_as["pressure"].values
means_as  = df_as["mean"].values
stds_as   = df_as["std"].values

labels_p = df_p["pressure"].values
means_p  = df_p["mean"].values
stds_p   = df_p["std"].values

fig, ax = plt.subplots(figsize=(2.4, 2.4))

ax.errorbar(
    labels_p,
    means_p,
    yerr=stds_p,
    marker='o',
    linestyle='--',
    color='#7c87b5',
    markersize=5,
    capsize=3,
    label=r"$a$-P",
    zorder=3
)

ax.errorbar(
    labels_as,
    means_as,
    yerr=stds_as,
    marker='o',
    linestyle='-',
    color='#cc5d5d',
    markersize=5,
    capsize=3,
    label=r"$a$-As",
    zorder=3
)

ax.set_xlabel("Pressure (GPa)")
ax.set_ylabel("Five-membered rings per atom")
ax.set_xlim(-1, 7)
ax.set_xticks(np.arange(0, 6.01, 2))
ax.set_ylim(0.16, 0.5)
ax.set_yticks(np.arange(0.2, 0.51, 0.1))
ax.yaxis.set_tick_params(which='both', labelbottom=True, bottom=True, zorder=10)
ax.xaxis.set_tick_params(which='both', labelbottom=True, bottom=True, zorder=10)
ax.legend()

plt.show()
```
