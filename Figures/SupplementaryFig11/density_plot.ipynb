```python
import numpy as np
import matplotlib.pyplot as plt

dens_md = np.loadtxt("densities_md.txt")
dens_rss = np.loadtxt("densities_rss.txt")

dens_md = np.atleast_1d(dens_md)
dens_rss = np.atleast_1d(dens_rss)
dens_md = dens_md[np.isfinite(dens_md)]
dens_rss = dens_rss[np.isfinite(dens_rss)]

fig, ax = plt.subplots(figsize=(3.5, 3))

md_centers = np.array([4.14, 4.44, 4.74, 5.04, 5.34])
bin_width = 0.11

counts_md = np.zeros_like(md_centers, dtype=int)
for i, c in enumerate(md_centers):
    left = c - bin_width / 2
    right = c + bin_width / 2
    counts_md[i] = np.sum((dens_md >= left) & (dens_md < right))

ax.bar(
    md_centers,
    counts_md,
    width=bin_width,
    alpha=0.7,
    color='C0',
    label='MD'
)

bins_rss = 40
counts_rss, edges_rss = np.histogram(dens_rss, bins=bins_rss, density=False)
centers_rss = 0.5 * (edges_rss[:-1] + edges_rss[1:])
width_rss = edges_rss[1] - edges_rss[0]

ax.bar(
    centers_rss,
    counts_rss,
    width=width_rss,
    alpha=0.5,
    color='C1',
    label='RSS'
)

ax.set_xlabel("Density (g cm$^{-3}$)")
ax.set_ylabel("Count")
ax.set_xlim(3, 6.5)

ymax = max(counts_md.max() if counts_md.size else 0,
           counts_rss.max() if counts_rss.size else 0)
ax.set_ylim(0, max(10, int(ymax * 1.15)))

ax.legend(frameon=False)

plt.show()
```
