```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("soap_atom_pca_data.csv")

mask_rss = df['source'] == 'rss_train.extxyz'
mask_amorph = df['source'] == 'amorphous.extxyz'

df_rss = df[mask_rss]
df_amorph = df[mask_amorph]

fig, ax = plt.subplots(figsize=(3.0, 3.0))

ax.scatter(df_rss['PC1'], df_rss['PC2'],
           c='#1f77b4',
           label='RSS',
           s=15,
           marker='o',
           zorder=4)

ax.scatter(df_amorph['PC1'], df_amorph['PC2'],
           c='#ff7f0e',
           label='Amorphous',
           s=12,
           marker='s',
           zorder=4)

ax.set_xlabel('PC1')
ax.set_ylabel('PC2')
ax.tick_params(axis='both', which='major')
ax.grid(False)
ax.legend(frameon=True)
ax.set_ylim(-25, 30)
ax.set_yticks(range(-20, 31, 10))
ax.set_xlim(-35, 30)
ax.set_xticks(range(-30, 31, 10))

plt.show()
```
