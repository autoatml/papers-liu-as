```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("soap_iteration_saturation_saturation_summary_consecutive.csv")

x = df.iloc[:,0].values
y = df.iloc[:,1].values

fig, ax = plt.subplots(figsize=(3.0, 3.0))

ax.scatter(x, y, s=30, c='#1f77b4', edgecolor='k', linewidth=0.3, zorder=4)
ax.plot(x, y, linestyle='--', linewidth=1.0, color='#1f77b4', zorder=3)

ax.set_ylim(-0.01, 0.18)
ax.set_yticks(np.arange(0, 0.181, 0.03))
ax.set_xlim(50, 1500)
ax.set_xticks(range(100, 1501, 300))

ax.set_xlabel('Number of configurations')
ax.set_ylabel('Jensen–Shannon divergence')

ax.tick_params(axis='both', which='major', labelsize=7)
ax.grid(False)

plt.show()
```
