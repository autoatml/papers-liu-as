```python
import matplotlib.pyplot as plt
import numpy as np

energy_error = {
    4: [0.01216591181469483, 0.024052358601766403, 0.020762769417453718, 0.02137399408816198],
    6: [0.006215861924740636, 0.012262851178816436, 0.01003788701377037, 0.011368529534215665],
    8: [0.008380817479824686, 0.01325723694747349, 0.012270406774200125, 0.013247828275929508],
    10: [0.023397017012442834, 0.033013830846633105, 0.03059707930634958, 0.032000841219867286],
}

force_error = {
    4: [0.12783266957352604, 0.11136800671337287, 0.11171635104670449, 0.10780684867813406],
    6: [0.11759337395049378, 0.10073468572070957, 0.09828052815169964, 0.09503688481518978],
    8: [0.12802774696621907, 0.10672483400694457, 0.10455066259458115, 0.10257015741596073],
    10: [0.1672875316291032, 0.15209626981975063, 0.15247506343374392, 0.14932918207785273],
}

x_atoms = np.array([256,512,1024])

colors=['#1b9e77', '#d95f02', '#7570b3', '#e7298a']
markers=['o','s','^','D']
cutoffs=[4,6,8,10]

fig_e, ax_e = plt.subplots(figsize=(2.4, 2.4))

for i, c in enumerate(cutoffs):
    y = np.array(energy_error[c][1:])
    y = y*1000
    ax_e.scatter(x_atoms, y,
                 color=colors[i], marker=markers[i], s=30,
                 edgecolor='k', linewidth=0.4, zorder=5,
                 label=f"cutoff = {c} Å")

    ax_e.plot(x_atoms, y, color=colors[i], linewidth=1.0,
                linestyle='--', dashes=(3,2), zorder=3)

ax_e.set_xlabel("Number of atoms")
ax_e.set_ylabel(r"Energy RMSE (meV $\mathrm{atom}^{-1}$)")
ax_e.set_xticks(x_atoms)
ax_e.set_xlim(x_atoms.min() - 30, x_atoms.max() + 30)

ax_e.set_ylim(5,45)
ax_e.set_yticks(np.arange(5,46,10))

ax_e.legend()
ax_e.grid(False)

plt.show()
```

```python
fig_f, ax_f = plt.subplots(figsize=(2.4, 2.4))

for i, c in enumerate(cutoffs):
    y = np.array(force_error[c][1:])
    ax_f.scatter(x_atoms, y,
                 color=colors[i], marker=markers[i], s=30,
                 edgecolor='k', linewidth=0.4, zorder=5,
                 label=f"cutoff = {c} Å")
    ax_f.plot(x_atoms, y, color=colors[i], linewidth=1.0,
                linestyle='--', dashes=(3,2), zorder=3)

ax_f.set_xlabel("Number of atoms")
ax_f.set_ylabel("Force RMSE (eV $\mathrm{\AA^{-1}}$)")
ax_f.set_xticks(x_atoms)
ax_f.set_xlim(x_atoms.min() - 30, x_atoms.max() + 30)

ax_f.set_ylim(0.08, 0.17)
ax_f.set_yticks(np.arange(0.08, 0.17, 0.02))

ax_f.legend()
ax_f.grid(False)

plt.show()
```
