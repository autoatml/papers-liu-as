```python
import matplotlib.pyplot as plt
import numpy as np

files = ['Sq_4A.txt', 'Sq_5A.txt', 'Sq_6A.txt', 'Sq_7A.txt', 'Sq_8A.txt']
labels = ["4 Å", 
          "5 Å",
          "6 Å",
          "7 Å",
          "8 Å",]
colors = ['blue', 'red', 'green', 'orange', 'purple']
linestyles = ['-', '--', '-', '--', '-']


fig, ax = plt.subplots(figsize=(2.4, 2.4))

for fname, lab, col, ls in zip(files, labels, colors, linestyles):
    x, y = np.loadtxt(fname, unpack=True)
    ax.plot(x, y, linestyle=ls, color=col, label=lab)
    
ax.xaxis.set_tick_params(which='both', labelbottom=True, bottom=True)
ax.yaxis.set_tick_params(which='both', labelbottom=True, bottom=True)

plt.xlabel(r"${Q}$ ($\mathrm{\AA^{-1}}$)")
plt.ylabel(r"Structure factor")

ax.set_xlim(-1.5, 16)
ax.set_xticks(np.arange(0, 16.1, 4))
ax.set_ylim(0, 2.0)
ax.set_yticks(np.arange(0, 2.1, 1))
ax.legend(frameon=False)

for i in range(0, 2, 1):
    ax.axhline(y=i, linestyle='--', linewidth=0.65, color='gray', zorder=0)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

files = ['Sq_4A.txt', 'Sq_5A.txt', 'Sq_6A.txt', 'Sq_7A.txt', 'Sq_8A.txt']
labels = ["4 Å", 
          "5 Å",
          "6 Å",
          "7 Å",
          "8 Å",]
colors = ['blue', 'red', 'green', 'orange', 'purple']
linestyles = ['-', '--', '-', '--', '-']

x1_exp, y1_exp = np.loadtxt('sq_exp.txt', unpack=True)

fig, ax = plt.subplots(figsize=(2.4, 2.4))

for i, label in enumerate(labels): 
    x, y = np.loadtxt(files[i], unpack=True)
    y_interp = np.interp(x1_exp, x, y)
    difference = y1_exp - y_interp
    if i == 2:
        ax.plot(x1_exp, difference, color=colors[i], label=label, linestyle=linestyles[i], zorder = 10)
    else: 
        ax.plot(x1_exp, difference, color=colors[i], label=label, linestyle=linestyles[i])
    
ax.axhline(0, linestyle='--', color='gray', linewidth=0.7)
plt.xlabel(r"${Q}$ ($\mathrm{\AA^{-1}}$)")
plt.ylabel(r"Difference in S(${Q}$)")
ax.set_xlim(0, 10)
ax.set_xticks(np.arange(0, 10.1, 2))
ax.set_ylim(-0.5, 0.5) 
ax.set_yticks(np.arange(-0.5, 0.6, 0.5))
ax.legend(frameon=False)

plt.show()
```
