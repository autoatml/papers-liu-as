```python
import os
import matplotlib.pyplot as plt
import numpy as np
from scipy.ndimage import gaussian_filter1d
import seaborn as sns
from matplotlib.colors import ListedColormap
import ase.io
import matplotlib as mpl
from matplotlib.ticker import ScalarFormatter
from matplotlib.ticker import FuncFormatter
import pandas as pd

mpl.rcParams['font.family'] = 'Arial'
plt.rcParams.update({
    'axes.labelsize': 8,
    'axes.facecolor': 'w',
    'xtick.labelsize': 6,
    'ytick.labelsize': 6,
    'legend.fontsize': 6,
    'legend.fancybox': False,
    'legend.edgecolor': 'k',
    'legend.borderaxespad': 1.5,
    'lines.linewidth': 1.5,
    'axes.linewidth': 0.65,
    'xtick.direction': 'in',
    'xtick.major.size': 5,
    'xtick.major.width': 0.65,
    'xtick.minor.size': 3,
    'xtick.minor.width': 0.65,
    'ytick.direction': 'in',
    'ytick.major.size': 5,
    'ytick.major.width': 0.65,
    'ytick.minor.size': 3,
    'ytick.minor.width': 0.65,
})
plt.rcParams['mathtext.fontset'] = 'custom'
plt.rcParams['mathtext.rm'] = 'Arial'
plt.rcParams['mathtext.it'] = 'Arial:italic'
plt.rcParams['mathtext.bf'] = 'Arial:bold'

x1_exp, y1_exp = np.loadtxt('sq_exp.txt', unpack=True)

exp_height = np.array([0.54675, 1.46066])
r2scan_height = np.array([0.5440382934982888, 1.4429319578060629])
r2scan_height_errors = (r2scan_height - exp_height) / exp_height * 100
r2scan_rvv10_height = np.array([0.4115181903686742, 1.61352586029636])
r2scan_rvv10_height_errors = (r2scan_rvv10_height-exp_height) / exp_height * 100
tpss_height = np.array([0.493072545867146, 1.6387528747908904])
tpss_height_errors = (tpss_height-exp_height) / exp_height * 100

data = np.vstack([
    [r2scan_height_errors[0], r2scan_height_errors[1]],
    [r2scan_rvv10_height_errors[0], r2scan_rvv10_height_errors[1]],
    [tpss_height_errors[0], tpss_height_errors[1]]
])

labels = ['r2SCAN', 'r2SCAN-rVV10', 'TPSS']
colors = ['#cc5c5c', '#c7cc5d', '#b65dcc']
colors = ["#66c2a5", "#fc8d62", "#8da0cb"]
colors = ["#aec7e8", "#ffbb78", "#98df8a"]
colors = ["#9ecae1", "#6baed6", "#2171b5"]
colors = ["#4C72B0", "#DD8452", "#55A868"]
colors = ["#5B8FA8", "#C17C59", "#7A9A61"]
colors = ["#C17F91", "#7c87b5", "#88A67C"]

n_groups = data.shape[1]   
n_methods = data.shape[0] 

group_x = np.arange(n_groups)  

bar_width = 0.2
offsets = (np.arange(n_methods) - (n_methods - 1) / 2) * bar_width

fig, ax = plt.subplots(figsize=(2.0, 2.5))

for i in range(n_methods):
    ax.bar(group_x + offsets[i], data[i], width=bar_width, label=labels[i], color=colors[i])

ax.set_xticks(group_x)
ax.set_xticklabels(['FSDP height', 'SSDP height'])

ax.set_ylabel('Relative error (%)')
# ax.set_xlabel('Height type')
ax.legend()

# give some space between the two groups visually by setting x-limits a bit wider
left = group_x[0] - 0.6
right = group_x[-1] + 0.6
ax.set_xlim(left, right)

ax.axhline(0, color='black', linewidth=0.6)

ax.set_ylim(-30, 20)
ax.set_yticks(np.arange(-30, 21, 10))

plt.savefig("high_res_mace_error.png", dpi=1000, format='png', bbox_inches='tight', transparent=False)
plt.savefig("high_res_mace_error.svg", dpi=1000, format='svg', bbox_inches='tight', transparent=True)

plt.show()
```
