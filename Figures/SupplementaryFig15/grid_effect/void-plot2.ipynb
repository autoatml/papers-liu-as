```python
import matplotlib.pyplot as plt
import pandas as pd

grids = [0.09, 0.19, 0.29]

fig, axes = plt.subplots(nrows=3, ncols=1, figsize=(3, 4), sharex=True)

max_y = 0
dfs = {}

for grid in grids:
    df = pd.read_csv(f"histogram_pc1_{grid}_2.9_5.csv")
    bin_width = df["bin_center"].diff().median()
    area = (df["count"] * bin_width).sum()
    df["count_norm"] = df["count"] / area
    dfs[grid] = df
    max_y = max(max_y, df["count_norm"].max())

ylim = max_y * 1.1

for i, grid in enumerate(grids):
    df = dfs[grid]

    axes[i].plot(df["bin_center"], df["count_norm"], '-', color="#C497b2", label=f"{grid} Å")
    axes[i].fill_between(df["bin_center"], df["count_norm"], color="#C497b2", alpha=0.1)

    axes[i].set_ylim(0, ylim)
    axes[i].set_xlim(-25, 25)

    axes[i].set_yticks([])
    axes[i].set_yticklabels([])

    axes[i].legend(loc="upper right", frameon=False)

plt.subplots_adjust(hspace=0)
axes[2].set_xlabel("Position (Å)")

fig.text(0.06, 0.5, "Count", va="center", rotation="vertical", fontsize=8)

plt.show()
```
