```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

work_dir = Path('.')  

color_map = {
    'aAs': '#7c87b5',
    'blackAs': '#000000',
    'greyAs': '#929292',
    'yellowAs': '#f2d600',
    'violetAs': '#a000a0',
    'fibrousAs': '#53006D'
}

for npz in sorted(work_dir.glob('*_cohpdata.npz')):
    data = np.load(npz, allow_pickle=True)

    name = str(data['name'])
    key = npz.stem.replace('_cohpdata', '')
    color = color_map.get(key, None)

    dos_energies = data['dos_energies']
    dos_dens_up  = data['dos_dens_up']
    cohp_energies = data['cohp_energies']
    cohp_broadened = data['cohp_broadened']
    cobi_energies = data['cobi_energies']
    cobi_broadened = data['cobi_broadened']

    fig, ax = plt.subplots(1, 3, figsize=(2.4, 2.4), sharey=True)

    ax[0].fill_between(dos_dens_up, dos_energies, color=color, alpha=0.3)
    ax[0].plot(dos_dens_up, dos_energies, color=color)
    ax[0].axhline(0, ls='--', lw=0.5, c='k')
    ax[0].set_xlabel("DOS")
    ax[0].set_ylabel("Energy (eV)")
    ax[0].set_xticks([])
    ax[0].set_yticks([-6, -4, -2, 0, 2, 4])

    ax[1].fill_between(-cohp_broadened, cohp_energies, color=color, alpha=0.3)
    ax[1].plot(-cohp_broadened, cohp_energies, color=color)
    ax[1].axhline(0, ls='--', lw=0.5, c='k')
    ax[1].axvline(0, lw=0.5, c='k')
    ax[1].set_xlabel("–COHP")
    ax[1].set_xlim(-1, 1)
    ax[1].set_ylim(-6, 5)

    ax[2].fill_between(cobi_broadened, cobi_energies, color=color, alpha=0.3)
    ax[2].plot(cobi_broadened, cobi_energies, color=color)
    ax[2].axhline(0, ls='--', lw=0.5, c='k')
    ax[2].axvline(0, lw=0.5, c='k')
    ax[2].set_xlabel("COBI")
    ax[2].set_xlim(-0.5, 0.5)

    plt.suptitle(name, fontsize=9)
    outname = f"{key}_cohpplots.png"
    plt.savefig(outname, dpi=300, bbox_inches='tight')
    plt.close(fig)

    print(f"Plotted {outname}")
```
